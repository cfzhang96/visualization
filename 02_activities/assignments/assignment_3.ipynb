```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
#load TTC delay dataset
df = pd.read_csv("TTC LRT Delays.csv")
```

```python
# Visualization 1:
# Top 10 LRT stations by average delay

station_delay = (
    df.groupby("Station")["Min Delay"]
    .mean()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(8, 4))
station_delay.plot(kind="bar")
plt.title("Top 10 LRT Stations by Average Delay")
plt.xlabel("Station")
plt.ylabel("Average Delay (minutes)")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()

```

> What information or message are you trying to convey with your visualization? 
  The visualization highlights the LRT stations with the highest average delay times.
  The key message is that delays are not evenly distributed across stations—instead, a small number of stations experience disproportionately high average delays, suggesting potential operational bottleneck

> What aspects of design did you consider when making your visualization? How did you apply them? With what elements of your plots?  
  -Chart type: I used barplot to compare discrete categories (stations)
  -Reduction of clutter: Only top10 was shown
  -Labeling:Clear axis labels and units
  -Test rotation: Stations names were rotated to avoid overlap

```python
# Visualization 2:
# Average LRT delay by day of week

day_order = [
    "Monday", "Tuesday", "Wednesday",
    "Thursday", "Friday", "Saturday", "Sunday"
]

day_delay = (
    df.groupby("Day")["Min Delay"]
    .mean()
    .reindex(day_order)
)

plt.figure(figsize=(7, 4))
plt.plot(day_delay.index, day_delay.values, marker="o")
plt.title("Average LRT Delay by Day of Week")
plt.xlabel("Day of Week")
plt.ylabel("Average Delay (minutes)")
plt.tight_layout()
plt.show()
```

  -Ordering: Days were explicitly ordered from Monday to Sunday.
